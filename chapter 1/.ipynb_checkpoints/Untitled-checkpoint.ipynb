```python
from sklearn.datasets import load_iris
iris = load_iris(as_frame=True)
df = iris.data
df.head()
```

```python
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
```

```python
#feature selection
x = df[['petal length (cm)']]
y = df['sepal length (cm)']
```

```python
x_train,x_test,y_train, y_test = train_test_split(x,y, test_size = 0.2, random_state = 42)
# train the model
model = LinearRegression()
model.fit(x_train,y_train)
```

```python
# prediction and evaluation
y_pred = model.predict(x_test)
print(y_pred)
```

```python
print("r2 score: ", r2_score(y_test,y_pred))
print("MSE: ", mean_squared_error(y_test,y_pred))
```

```python
plt.scatter(x_test,y_test, color = 'blue', label = 'Actual Data')
plt.plot(x_test,y_pred, color= 'red', label= 'regression line')
plt.xlabel('petal length (cm)')
plt.ylabel('sepal length (cm)')
plt.title("simple linear regression")
plt.grid()
plt.show()
```

```python
import plotly.graph_objects as go

# 1. Scatter points banayein (Actual data with hover info)
fig = go.Figure()

fig.add_trace(go.Scatter(
    x=x_test['petal length (cm)'],
    y=y_test,
    mode='markers',
    name='Actual Data',
    marker=dict(size=9, color='blue'),
    hovertemplate='<b>Petal:</b> %{x:.2f} cm<br><b>Sepal:</b> %{y:.2f} cm<extra></extra>'
))

# 2. Regression line add karein
fig.add_trace(go.Scatter(
    x=x_test['petal length (cm)'],
    y=y_pred,
    mode='lines',
    name='Regression Line',
    line=dict(color='red', width=2)
))

# 3. Layout set karein
fig.update_layout(
    title='Interactive Linear Regression (Hover & Click points)',
    xaxis_title='Petal Length (cm)',
    yaxis_title='Sepal Length (cm)',
    template='plotly_white'
)

fig.show()
```
